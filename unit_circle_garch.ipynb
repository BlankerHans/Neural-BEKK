# GARCH and the Unit Circle

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
x = np.linspace(-1, 1, 1000)
```

Unit circle: $\sqrt{x^2 + y^2} = 1$

Sinus: $\sin(\theta) = \frac{\text{Gegenkathete}}{\text{Hypotenuse}}$

Sinus: $\cos(\theta) = \frac{\text{Ankathete}}{\text{Hypotenuse}}$ <- Ankathete am Winkel anliegende Linie

```python
y_pos = np.sqrt(1-x**2)
y_neg = -np.sqrt(1-x**2)
```

```python
plt.plot(x, y_pos)
plt.plot(x, y_neg)
plt.axis('equal')



plt.show() 
```

```python
# Einheitskreis
t = np.linspace(0, 2*np.pi, 1000)
x_circle = np.cos(t)
y_circle = np.sin(t)

# Beispielwinkel
theta = np.pi / 4  # 45°
x = np.cos(theta)
y = np.sin(theta)

plt.figure(figsize=(5, 5))

# Kreis
plt.plot(x_circle, y_circle)

# Radius (Winkel)
plt.plot([0, x], [0, y])

# Dreieck (cos und sin)
plt.plot([0, x], [0, 0])   # cos(theta)
plt.plot([x, x], [0, y])   # sin(theta)

# Punkt auf dem Kreis
plt.scatter(x, y)

# Beschriftungen
plt.text(x/2, -0.05, r'$\cos(\theta)$', ha='center')
plt.text(x + 0.05, y/2, r'$\sin(\theta)$', va='center')
plt.text(x/2, y/2, r'$1$', ha='center')

# Achsen
plt.axhline(0)
plt.axvline(0)
plt.axis('equal')
plt.xlim(-1.2, 1.2)
plt.ylim(-1.2, 1.2)

plt.title(r'Einheitskreis mit $\sin(\theta)$ und $\cos(\theta)$')
plt.show()
```

Allgemein gilt für einen Kreis mit Radius $r$ und einem Punkt $P(x,y)$ mit Winkel $\theta$ $\Rightarrow$ $\cos(\theta) = \frac{x}{r}$, da $x$ die Ankathete und $r$ die Hypotenuse.

Jeder Punkt auf dem Einheitskreis hat $P(x,y)$ mit $x=\cos(\theta)$ und $y=\sin(\theta)$, da die Hypotenuse innerhalb des Einheitskreises immer $r=1$ aufweist!

$-1 < \cos \theta <1$

- ändert sich der Radius des Kreise, dann ändert sich $x$ Automatisch, allgemein $x = r \cdot \cos(\theta)$
- der Winkel $\theta$ ist fix und damit hängt der $\cos$ und der $\sin$ nicht von der größe des Dreiecks ab
